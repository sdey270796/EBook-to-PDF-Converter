# Part III: **Detect the File Format**

## Extract File Name and Extension

```python
# Split file name and extension
file_name, file_extension = os.path.splitext(input_file)

# Remove the leading '.'
file_extension = file_extension.lower().replace(".", "")

print(f"File Name      : {file_name}")
print(f"File Extension : {file_extension}")
```

## Define Supported Formats

```python
supported_formats = {
    "epub",
    "mobi",
    "azw",
    "azw3",
    "fb2",
    "txt",
    "html",
    "htm",
    "rtf",
    "docx",
    "pdf"
}
```

## Validate the Format

```python
if file_extension in supported_formats:
    print(f"✅ '{file_extension}' is a supported format.")
else:
    raise ValueError(f"❌ '{file_extension}' is not a supported ebook format.")
```

## Create the Output PDF Name

```python
output_file = f"{file_name}.pdf"
output_path = os.path.join("/content", output_file)

print(f"Output PDF : {output_path}")
```

# Part VII: **Build the Calibre Command Dynamically**

## Create the Base Command

```python
command = [
    "ebook-convert",
    input_path,
    output_path
]
```

### Add Paper Size

```python
command.extend([
    "--paper-size", paper_size
])
```

### Add Margins

```python
command.extend([
    "--pdf-page-margin-top", top_margin,
    "--pdf-page-margin-bottom", bottom_margin,
    "--pdf-page-margin-left", left_margin,
    "--pdf-page-margin-right", right_margin
])
```

### Add Font Sizes

```python
command.extend([
    "--base-font-size", base_font_size,
    "--pdf-default-font-size", pdf_font_size
])
```

### Paragraph Spacing

```python
if remove_spacing.lower() == "y":
    command.append("--remove-paragraph-spacing")

if insert_blank_line.lower() == "y":
    command.extend([
        "--insert-blank-line",
        "--insert-blank-line-size", blank_line_size
    ])
```

### Text Justification

```python
if justify.lower() == "y":

    css = """
    body {
        text-align: justify;
    }
    """

    css_path = "/content/custom_style.css"

    with open(css_path, "w") as f:
        f.write(css)

    command.extend([
        "--extra-css", css_path
    ])
```

## Display the Final Command

```python
print("Generated Command:\n")

print(" ".join(command))
```

# Part VIII: **Customized Conversion**

## Display Application Header

```python
print("=" * 60)
print("      Universal eBook to PDF Converter v1.5")
print("=" * 60)
print(f"📖 Input File : {input_file}")
print(f"📄 Output PDF : {output_file}")
print("=" * 60)
```

## Start Conversion

```python
import subprocess
import os
import threading
import time
import sys
```

```python
# ----------------------------------------------------
# Progress Bar Animation
# ----------------------------------------------------

progress = 0
conversion_finished = False


def progress_bar():

    global progress, conversion_finished

    status_messages = [
        "Initializing converter...",
        "Reading eBook...",
        "Processing chapters...",
        "Formatting pages...",
        "Embedding fonts...",
        "Rendering PDF...",
        "Optimizing layout...",
        "Finalizing PDF..."
    ]

    msg_index = 0

    while not conversion_finished:

        # Increase gradually until 95%
        if progress < 95:
            progress += 1

        filled = int(progress / 5)
        bar = "█" * filled + "░" * (20 - filled)

        print(
            f"\r[{bar}] {progress:3d}%   {status_messages[msg_index]}",
            end=""
        )

        if progress % 12 == 0:
            msg_index = (msg_index + 1) % len(status_messages)

        time.sleep(0.20)

    # Jump to 100%
    progress = 100
    bar = "█" * 20

    print(
        f"\r[{bar}] 100%   Conversion Complete!               "
    )
```

```python
'''
# Required for Google Colab
env = os.environ.copy()
env["QTWEBENGINE_DISABLE_SANDBOX"] = "1"
'''
```

```python
print("\n⏳ Converting eBook to PDF...\n")
```

```python
# ----------------------------------------------------
# Start Timer
# ----------------------------------------------------

start_time = time.time()

# Start Progress Thread
thread = threading.Thread(target=progress_bar)
thread.start()

try:

    result = subprocess.run(
        command,
        capture_output=True,
        text=True,
        check=True,
        env=env
    )

except subprocess.CalledProcessError as e:

    conversion_finished = True
    thread.join()

    print("\n\nConversion Failed!\n")
    print(e.stderr)

    raise

# Stop Animation
conversion_finished = True
thread.join()
```

```python
# ----------------------------------------------------
# Summary
# ----------------------------------------------------
elapsed = time.time() - start_time
```

```python
print("\n")
print("=" * 60)

if os.path.exists(output_path):

    size = os.path.getsize(output_path) / (1024 * 1024)

    print("✅ Conversion Successful")
    print(f"📄 Output File : {output_file}")
    print(f"📦 File Size   : {size:.2f} MB")
    print(f"⏱ Time Taken   : {elapsed:.2f} seconds")

else:

    print("❌ PDF was not generated.")

print("=" * 60)
```
